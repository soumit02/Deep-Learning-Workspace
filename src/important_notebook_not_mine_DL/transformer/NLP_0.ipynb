DEMO 01 — Tokenization with NLTK
=================================

```python
"""
Shows word tokenization, sentence tokenization, and the difference
between a naive split() and NLTK's smart word_tokenize().

NLTK downloads required (auto-downloaded on first run):
    punkt_tab, punkt
"""

import nltk

# Auto-download required NLTK data (safe to run multiple times)
print("Downloading NLTK data packages...")
nltk.download("punkt",       quiet=True)
nltk.download("punkt_tab",   quiet=True)
print("✅ NLTK data ready\n")

from nltk.tokenize import (
    word_tokenize,       # Smart word tokenizer (handles punctuation)
    sent_tokenize,       # Sentence tokenizer
    TweetTokenizer,      # Good for social media text
    MWETokenizer,        # Multi-word expressions ("New York" as one token)
)
```

```python
# ── SECTION 1: Word Tokenization ─────────────────────────────────
print("=" * 60)
print("SECTION 1: Word Tokenization")
print("=" * 60)

sample_text = "The quick brown fox doesn't jump over the lazy dog! It ran away at 3.5 mph..."

print(f"\nOriginal text:\n  '{sample_text}'\n")

# Method 1: Naive Python split — WRONG way
naive_tokens = sample_text.split()
print(f"1. Naive split()  → {len(naive_tokens)} tokens")
print(f"   {naive_tokens}\n")
print("   ⚠️  Problem: 'dog!' mixed with punctuation\n")

# Method 2: NLTK word_tokenize — CORRECT way
nltk_tokens = word_tokenize(sample_text)
print(f"2. word_tokenize()  → {len(nltk_tokens)} tokens")
print(f"   {nltk_tokens}\n")
print("   ✅ Punctuation separated. \"doesn't\" → [\"does\", \"n't\"] (smart!)\n")
```

```python
# ── SECTION 2: Sentence Tokenization ─────────────────────────────
print("=" * 60)
print("SECTION 2: Sentence Tokenization")
print("=" * 60)

multi_sent = (
    "Dr. Smith works at N.Y.U. He published a paper in 2024. "
    "The paper was titled 'NLP is amazing!' and covered tokenization. "
    "Mr. Brown reviewed it; he gave it 9.5/10."
)

print(f"\nMulti-sentence text:\n  '{multi_sent}'\n")

sentences = sent_tokenize(multi_sent)
print(f"sent_tokenize() found {len(sentences)} sentences:\n")
for i, sent in enumerate(sentences, 1):
    print(f"  Sentence {i}: {sent}")

print("\n  ✅ 'Dr.' and 'N.Y.U.' NOT treated as sentence ends — smart!")
print("  ✅ '9.5/10' NOT treated as sentence end")
```

```python
# ── SECTION 3: Tweet Tokenizer ────────────────────────────────────
print("\n" + "=" * 60)
print("SECTION 3: Tweet Tokenizer (social media text)")
print("=" * 60)

tweet = "OMG I LOVE this!! 😍 #NLP #Python @BrainStation23 check this url: https://t.co/xyz"
print(f"\nTweet text:\n  '{tweet}'\n")

tweet_tokenizer = TweetTokenizer(preserve_case=False, strip_handles=True, reduce_len=True)
tweet_tokens = tweet_tokenizer.tokenize(tweet)
print(f"TweetTokenizer → {len(tweet_tokens)} tokens:")
print(f"  {tweet_tokens}")
print("\n  ✅ Handles hashtags, mentions, emojis, URLs specially")
```

```python
# ── SECTION 4: Comparison Summary ────────────────────────────────
print("\n" + "=" * 60)
print("SECTION 4: Token Count Comparison")
print("=" * 60)

test = "Running foxes don't eat quickly. They're beautiful!"
print(f"\nTest text: '{test}'")
print(f"\n  split()          : {len(test.split())} tokens → {test.split()}")
print(f"  word_tokenize()  : {len(word_tokenize(test))} tokens → {word_tokenize(test)}")

print("\n" + "=" * 60)
print("✅ Demo 01 complete — NLTK Tokenization")
print("=" * 60)
```

DEMO 02 — Stopword Removal with NLTK and spaCy
================================================

```python
"""
Shows how to identify and filter stopwords using both libraries.
Demonstrates why stopword removal matters for NLP tasks.
"""

import nltk
nltk.download("punkt",     quiet=True)
nltk.download("punkt_tab", quiet=True)
nltk.download("stopwords", quiet=True)

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
import spacy
```

```python
print("Loading spaCy model...")
nlp = spacy.load("en_core_web_sm")
print("✅ spaCy loaded\n")
```

```python
# ── SECTION 1: What are stopwords? ───────────────────────────────
print("=" * 60)
print("SECTION 1: What are English Stopwords?")
print("=" * 60)

nltk_sw = set(stopwords.words("english"))
spacy_sw = nlp.Defaults.stop_words

print(f"\n  NLTK  English stopwords : {len(nltk_sw)} words")
print(f"  spaCy English stopwords : {len(spacy_sw)} words")

# Show some examples
print(f"\n  Sample NLTK  stopwords : {sorted(list(nltk_sw))[:20]}")
print(f"  Sample spaCy stopwords : {sorted(list(spacy_sw))[:20]}")

# Words in spaCy but not NLTK
only_in_spacy = spacy_sw - {w.lower() for w in nltk_sw}
print(f"\n  Words in spaCy but NOT in NLTK ({len(only_in_spacy)} extra):")
print(f"  {sorted(list(only_in_spacy))[:25]}")
```

```python
# ── SECTION 2: Removing stopwords with NLTK ──────────────────────
print("\n" + "=" * 60)
print("SECTION 2: Stopword Removal with NLTK")
print("=" * 60)

text = (
    "Machine learning is a subset of artificial intelligence. "
    "It enables systems to automatically learn and improve from experience. "
    "The algorithms are trained on large datasets to make predictions."
)
print(f"\nOriginal text:\n  '{text}'\n")

# Tokenize
tokens = word_tokenize(text.lower())
print(f"After tokenization ({len(tokens)} tokens):")
print(f"  {tokens}\n")

# Filter stopwords + punctuation
nltk_filtered = [
    token for token in tokens
    if token.isalpha() and token not in nltk_sw
]
print(f"After NLTK stopword removal ({len(nltk_filtered)} tokens):")
print(f"  {nltk_filtered}")

removed_count = len([t for t in tokens if t.lower() in nltk_sw])
print(f"\n  → Removed {removed_count} stopwords ({removed_count/len(tokens)*100:.1f}% of tokens)")
```

```python
# ── SECTION 3: Removing stopwords with spaCy ─────────────────────
print("\n" + "=" * 60)
print("SECTION 3: Stopword Removal with spaCy")
print("=" * 60)

doc = nlp(text)

print(f"\nspaCy token analysis:")
print(f"  {'Token':<20} {'is_stop':<10} {'is_alpha':<10} {'pos_'}")
print(f"  {'-'*55}")
for token in doc:
    print(f"  {token.text:<20} {str(token.is_stop):<10} {str(token.is_alpha):<10} {token.pos_}")

spacy_filtered = [
    token.text.lower()
    for token in doc
    if not token.is_stop and token.is_alpha
]
print(f"\nAfter spaCy stopword removal ({len(spacy_filtered)} tokens):")
print(f"  {spacy_filtered}")
```

```python
# ── SECTION 4: NLTK vs spaCy comparison ─────────────────────────
print("\n" + "=" * 60)
print("SECTION 4: NLTK vs spaCy Stopword Comparison")
print("=" * 60)

test_sentence = "The running fox is quickly jumping over a lazy dog near the river."
test_sentence1 = "Home alone is a good movie"
print(f"\nTest sentence: '{test_sentence1}'\n")

# NLTK
nltk_tokens = word_tokenize(test_sentence1.lower())
nltk_result = [t for t in nltk_tokens if t.isalpha() and t not in nltk_sw]

# spaCy
spacy_doc = nlp(test_sentence1)
spacy_result = [t.text.lower() for t in spacy_doc if not t.is_stop and t.is_alpha]

print(f"  NLTK  result ({len(nltk_result)} tokens) : {nltk_result}")
print(f"  spaCy result ({len(spacy_result)} tokens) : {spacy_result}")
```

```python
# ── SECTION 5: Multilingual stopwords (NLTK) ─────────────────────
print("\n" + "=" * 60)
print("SECTION 5: Multilingual Stopwords (NLTK)")
print("=" * 60)

languages = ["english", "french", "german", "arabic", "spanish", "bengali"]
for lang in languages:
    try:
        sw = stopwords.words(lang)
        sample = sorted(sw)[:5]
        print(f"  {lang:<12}: {len(sw):>3} stopwords — e.g. {sample}")
    except Exception as e:
        print(f"  {lang:<12}: not available ({e})")
```

```python
# ── SECTION 6: When NOT to remove stopwords ───────────────────────
print("\n" + "=" * 60)
print("SECTION 6: When NOT to Remove Stopwords")
print("=" * 60)

print("""
  ❌ DON'T remove stopwords for:
     • Sentiment analysis: "not good" → removing "not" → "good" (WRONG!)
     • BERT/GPT models: these models need full context including stopwords
     • Machine translation: every word matters for grammar
     • QA / chatbots: "Is this correct?" — all words needed

  ✅ DO remove stopwords for:
     • TF-IDF / bag of words features
     • Topic modelling (LDA)
     • Document classification with classic ML
     • Text summarization pre-processing
     • Keyword extraction
""")

print("=" * 60)
print("✅ Demo 02 complete — Stopword Removal")
print("=" * 60)
```

DEMO 03 — Stemming with NLTK
===========================

```python
"""
Demonstrates Porter, Snowball, and Lancaster stemmers.
Shows the difference between all three algorithms on
identical inputs, and compares stemming vs lemmatization.

NLTK downloads (auto-downloaded on first run):
    wordnet, averaged_perceptron_tagger_eng
"""

import nltk
nltk.download("wordnet",    quiet=True)
nltk.download("punkt",      quiet=True)
nltk.download("punkt_tab",  quiet=True)
nltk.download("stopwords",  quiet=True)
nltk.download("averaged_perceptron_tagger_eng", quiet=True)

from nltk.stem import PorterStemmer, SnowballStemmer, LancasterStemmer
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet, stopwords
from nltk.tokenize import word_tokenize
```

```python
porter    = PorterStemmer()
snowball  = SnowballStemmer("english")
lancaster = LancasterStemmer()
lemmatizer = WordNetLemmatizer()
sw = set(stopwords.words("english"))

print("✅ All stemmers loaded\n")


# ─── helpers ──────────────────────────────────────────────────────
def get_wordnet_pos(tag: str) -> str:
    m = {"J": wordnet.ADJ, "V": wordnet.VERB, "N": wordnet.NOUN, "R": wordnet.ADV}
    return m.get(tag[0], wordnet.NOUN)


def lemmatize_pos(word: str, pos_tag: str) -> str:
    return lemmatizer.lemmatize(word, get_wordnet_pos(pos_tag))
```

```python
# ═══════════════════════════════════════════════════════════════════
# SECTION 1: Basic stemmer comparison
# ═══════════════════════════════════════════════════════════════════
print("=" * 70)
print("SECTION 1: Porter vs Snowball vs Lancaster — Word by Word")
print("=" * 70)

test_words = [
    "running",   "studies",   "beautiful",  "generously",
    "connection","flies",     "wolves",     "went",
    "happily",   "maximum",   "caring",     "argued",
    "caresses",  "ponies",    "cats",       "trouble",
    "matting",   "mating",    "meetings",   "rats",
]

print(f"\n{'Word':<18} {'Porter':<18} {'Snowball':<18} {'Lancaster'}")
print(f"  {'-'*65}")
for w in test_words:
    p = porter.stem(w)
    s = snowball.stem(w)
    l = lancaster.stem(w)
    # flag words where algorithms disagree or produce non-words
    print(f"  {w:<18} {p:<18} {s:<18} {l}")
```

```python
# ═══════════════════════════════════════════════════════════════════
# SECTION 3: Multilingual Snowball
# ═══════════════════════════════════════════════════════════════════
print("=" * 70)
print("SECTION 3: Snowball Multilingual Support")
print("=" * 70)

print(f"\nSupported languages: {SnowballStemmer.languages}\n")

multilingual = [
    ("english",    ["running", "studies", "happiness"]),
    ("french",     ["manager",  "mangeons", "mangez"]),
    ("german",     ["laufen",  "läuft",   "gelaufen"]),
    ("spanish",    ["correr",  "corriendo","corremos"]),
    ("arabic",     ["درس",     "يدرس",    "دراسة"]),
]

for lang, words in multilingual:
    try:
        stemmer = SnowballStemmer(lang)
        stems   = [stemmer.stem(w) for w in words]
        print(f"  {lang:<12} {words} → {stems}")
    except Exception as e:
        print(f"  {lang:<12} → Error: {e}")

print("\n  ✅ Snowball supports 13+ languages — use it for multilingual NLP!")
```

```python
# ═══════════════════════════════════════════════════════════════════
# SECTION 4: Stemming on a full sentence
# ═══════════════════════════════════════════════════════════════════
print("\n" + "=" * 70)
print("SECTION 4: Full Sentence Stemming Pipeline")
print("=" * 70)

sentence = "The running scientists quickly studied better machine learning methods across multiple laboratories."
print(f"\nOriginal: '{sentence}'\n")

# Tokenize + lowercase
tokens = word_tokenize(sentence.lower())
# Remove stopwords + punctuation
clean_tokens = [t for t in tokens if t.isalpha() and t not in sw]

print(f"Tokens (no stopwords): {clean_tokens}\n")
print(f"{'Token':<18} {'Porter':<18} {'Snowball':<18} {'Lancaster'}")
print(f"  {'-'*65}")
for t in clean_tokens:
    p = porter.stem(t)
    s = snowball.stem(t)
    l = lancaster.stem(t)
    print(f"  {t:<18} {p:<18} {s:<18} {l}")

porter_result   = [porter.stem(t)   for t in clean_tokens]
snowball_result = [snowball.stem(t)  for t in clean_tokens]
lanc_result     = [lancaster.stem(t) for t in clean_tokens]

print(f"\nPorter   result: {porter_result}")
print(f"Snowball result: {snowball_result}")
print(f"Lancaster result: {lanc_result}")
```

DEMO 04 — Lemmatization with NLTK and spaCy
=============================================

```python
"""
Demonstrates lemmatization using both libraries.
Shows the critical difference from stemming.
Highlights how POS tagging improves accuracy.
"""

import nltk
nltk.download("wordnet",    quiet=True)
nltk.download("averaged_perceptron_tagger_eng", quiet=True)
nltk.download("punkt",     quiet=True)
nltk.download("punkt_tab", quiet=True)

from nltk.stem import WordNetLemmatizer, PorterStemmer, SnowballStemmer
from nltk.corpus import wordnet
from nltk.tokenize import word_tokenize
import nltk
import spacy
```

```python
print("Loading spaCy model...")
nlp = spacy.load("en_core_web_sm")
lemmatizer = WordNetLemmatizer()
porter = PorterStemmer()
snowball = SnowballStemmer("english")
print("✅ Models loaded\n")
```

```python
# ── SECTION 1: Stemming vs Lemmatization ─────────────────────────
print("=" * 65)
print("SECTION 1: Stemming vs Lemmatization — Side by Side")
print("=" * 65)

words = [
    "running", "studies", "better", "went", "corpora",
    "wolves", "geese", "happily", "caring", "argued",
    "beautiful", "flies", "jumping", "lazily", "worse",
]

print(f"\n  {'Word':<15} {'Porter Stem':<18} {'Snowball Stem':<18} {'NLTK Lemma':<18} {'spaCy Lemma'}")
print(f"  {'-'*80}")

for word in words:
    porter_stem   = porter.stem(word)
    snowball_stem = snowball.stem(word)
    nltk_lemma    = lemmatizer.lemmatize(word)
    spacy_token   = nlp(word)[0]
    spacy_lemma   = spacy_token.lemma_

    print(f"  {word:<15} {porter_stem:<18} {snowball_stem:<18} {nltk_lemma:<18} {spacy_lemma}")

print("\n")
print("  ✅  Lemmatization gives proper dictionary words")
```

```python
# ── SECTION 2: POS-aware lemmatization (NLTK) ────────────────────
print("\n" + "=" * 65)
print("SECTION 2: POS-Aware Lemmatization with NLTK WordNet")
print("=" * 65)

print("""
  Without POS: lemmatizer.lemmatize('better') → 'better' (WRONG)
  With POS:    lemmatizer.lemmatize('better', pos=wordnet.ADJ) → 'good' ✅

  POS tag constants in WordNet:
    wordnet.NOUN = 'n'   → lemmatize('wolves') → 'wolf'
    wordnet.VERB = 'v'   → lemmatize('running') → 'run'
    wordnet.ADJ  = 'a'   → lemmatize('better') → 'good'
    wordnet.ADV  = 'r'   → lemmatize('happily') → 'happily'
""")

pos_examples = [
    ("better",  wordnet.ADJ,  "ADJ"),
    ("running", wordnet.VERB, "VERB"),
    ("wolves",  wordnet.NOUN, "NOUN"),
    ("happily", wordnet.ADV,  "ADV"),
    ("studies", wordnet.VERB, "VERB"),
    ("studies", wordnet.NOUN, "NOUN"),
]

print(f"  {'Word':<12} {'POS':<8} {'No POS':<15} {'With POS'}")
print(f"  {'-'*50}")
for word, pos, pos_name in pos_examples:
    no_pos   = lemmatizer.lemmatize(word)
    with_pos = lemmatizer.lemmatize(word, pos=pos)
    flag = "✅" if no_pos != with_pos else "  "
    print(f"  {word:<12} {pos_name:<8} {no_pos:<15} {with_pos} {flag}")

```

```python
# ── SECTION 3: spaCy POS-aware lemmatization ─────────────────────
print("\n" + "=" * 65)
print("SECTION 3: spaCy POS-Aware Lemmatization (Auto-detected)")
print("=" * 65)

print("\n  spaCy automatically detects POS and uses it for lemmatization")
print("  No manual POS mapping needed!\n")

sentences = [
    "The wolves were running across the beautiful snowy fields.",
    "She is studying better strategies for the upcoming studies.",
    "The geese flew south while the children argued about going.",
]

for sent in sentences:
    doc = nlp(sent)
    print(f"\n  Input: '{sent}'")
    print(f"  {'Token':<15} {'POS':<10} {'Lemma'}")
    print(f"  {'-'*40}")
    for token in doc:
        if token.is_alpha:
            flag = "✅" if token.text.lower() != token.lemma_ else "  "
            print(f"  {token.text:<15} {token.pos_:<10} {token.lemma_} {flag}")
    lemmas = [t.lemma_.lower() for t in doc if t.is_alpha and not t.is_stop]
    print(f"\n  Clean lemmas (no stopwords): {lemmas}")
```

```python
# ── SECTION 4: Full pipeline comparison ──────────────────────────

def get_wordnet_pos(treebank_tag: str) -> str:
  """Convert Penn Treebank POS tag to WordNet POS constant."""
  if treebank_tag.startswith("J"):
      return wordnet.ADJ
  elif treebank_tag.startswith("V"):
      return wordnet.VERB
  elif treebank_tag.startswith("N"):
      return wordnet.NOUN
  elif treebank_tag.startswith("R"):
      return wordnet.ADV
  else:
      return wordnet.NOUN  # Default to noun


def lemmatize_with_pos_nltk(text: str) -> list[str]:
  """Full NLTK lemmatization pipeline with automatic POS detection."""
  tokens = word_tokenize(text.lower())
  pos_tags = nltk.pos_tag(tokens)
  return [
      lemmatizer.lemmatize(token, get_wordnet_pos(pos))
      for token, pos in pos_tags
      if token.isalpha()
  ]

print("\n" + "=" * 65)
print("SECTION 4: Complete Pipeline Comparison on Real Text")
print("=" * 65)

review = (
    "The running shoes are better than expected. "
    "My feet were hurting after walking, but these beautifully designed shoes "
    "helped significantly. The children also loved wearing them."
)
print(f"\nReview: '{review}'\n")

# NLTK pipeline
nltk_result = lemmatize_with_pos_nltk(review)
print(f"  NLTK lemmas  ({len(nltk_result)} tokens):")
print(f"  {nltk_result}")

# spaCy pipeline
spacy_doc = nlp(review)
spacy_result = [t.lemma_.lower() for t in spacy_doc if t.is_alpha and not t.is_stop]
print(f"\n  spaCy lemmas ({len(spacy_result)} tokens):")
print(f"  {spacy_result}")

print("\n" + "=" * 65)
print("✅ Demo 03 complete — Lemmatization")
print("=" * 65)
```

Stemming vs Lemmatization — direct comparison

```python
compare_words = [
    ("running",    "VBG"),
    ("studies",    "NNS"),
    ("better",     "JJR"),
    ("went",       "VBD"),
    ("wolves",     "NNS"),
    ("happily",    "RB"),
    ("corpora",    "NNS"),
    ("argued",     "VBD"),
    ("beautifully","RB"),
    ("saw",        "VBD"),
]

print(f"\n{'Word':<16} {'POS':<8} {'Porter Stem':<18} {'Snowball Stem':<18} {'Lemma (POS-aware)':<20} {'Real word?'}")
print(f"  {'-'*85}")
for word, pos_tag in compare_words:
    p_stem  = porter.stem(word)
    s_stem  = snowball.stem(word)
    lemma   = lemmatize_pos(word, pos_tag)
    # check if lemma is different from stem (accuracy gain)
    gain = "🎯 More accurate" if p_stem != lemma else ""
    print(f"  {word:<16} {pos_tag:<8} {p_stem:<18} {s_stem:<18} {lemma:<20} {gain}")

print("""
  Key takeaways:
  • 'better' → Porter: 'better' (unchanged), Lemma with ADJ: 'good' 🎯
  • 'went'   → Porter: 'went'   (unchanged), Lemma with VERB: 'go'   🎯
  • 'corpora'→ Porter: 'corpora'(unchanged), Lemma with NOUN: 'corpus'🎯
  • 'saw'    → Porter: 'saw'    (same),      Lemma with VERB: 'see'  🎯
""")
```

Speed comparison

```python
import time

big_word_list = test_words * 500  # 10,000 words

def bench(name, fn, words):
    t0 = time.perf_counter()
    [fn(w) for w in words]
    return (time.perf_counter() - t0) * 1000

p_ms  = bench("Porter",    porter.stem,    big_word_list)
s_ms  = bench("Snowball",  snowball.stem,  big_word_list)
l_ms  = bench("Lancaster", lancaster.stem, big_word_list)

# Lemmatization (without POS — fair comparison)
lem_ms = bench("Lemmatizer", lemmatizer.lemmatize, big_word_list)

print(f"""
10,000 words processed

Porter     : {p_ms:.0f} ms
Snowball   : {s_ms:.0f} ms
Lancaster  : {l_ms:.0f} ms
Lemmatizer : {lem_ms:.0f} ms

Note:
• This benchmark measures WordNet lemmatization without POS tagging.
• A full lemmatization pipeline (POS tagging + lemmatization) is generally slower
  than stemming but produces more accurate base forms.
""")
```

```python
import time
import nltk
from nltk import pos_tag
from nltk.corpus import wordnet

big_word_list = test_words * 500  # 10,000 words


def bench(fn):
    t0 = time.perf_counter()
    fn()
    return (time.perf_counter() - t0) * 1000


# ---------------- Stemmers ----------------
p_ms = bench(lambda: [porter.stem(w) for w in big_word_list])
s_ms = bench(lambda: [snowball.stem(w) for w in big_word_list])
l_ms = bench(lambda: [lancaster.stem(w) for w in big_word_list])


# ---------------- Lemmatization with POS ----------------
def get_wordnet_pos(treebank_tag):
    if treebank_tag.startswith("J"):
        return wordnet.ADJ
    elif treebank_tag.startswith("V"):
        return wordnet.VERB
    elif treebank_tag.startswith("N"):
        return wordnet.NOUN
    elif treebank_tag.startswith("R"):
        return wordnet.ADV
    return wordnet.NOUN


def pos_lemmatize():
    tags = pos_tag(big_word_list)
    return [
        lemmatizer.lemmatize(word, get_wordnet_pos(tag))
        for word, tag in tags
    ]


lem_ms = bench(pos_lemmatize)


print(f"""
10,000 words processed:

Porter      : {p_ms:.0f} ms ({p_ms/10:.2f} ms / 1000 words)
Snowball    : {s_ms:.0f} ms ({s_ms/10:.2f} ms / 1000 words)
Lancaster   : {l_ms:.0f} ms ({l_ms/10:.2f} ms / 1000 words)
Lemmatizer  : {lem_ms:.0f} ms ({lem_ms/10:.2f} ms / 1000 words)

Porter is {lem_ms/p_ms:.2f}× faster than POS-aware lemmatization.
Snowball is {lem_ms/s_ms:.2f}× faster than POS-aware lemmatization.
Lancaster is {lem_ms/l_ms:.2f}× faster than POS-aware lemmatization.
""")
```

When to use which — decision guide

DEMO 06 — Complete NLP Preprocessing Pipeline
===============================================

```python
"""
Combines tokenization + stopword removal + lemmatization
using NLTK, spaCy, and HuggingFace — all in one file.

Shows the full journey from raw text to clean tokens
ready for machine learning models.
"""

import re
import nltk
nltk.download("punkt",       quiet=True)
nltk.download("punkt_tab",   quiet=True)
nltk.download("stopwords",   quiet=True)
nltk.download("wordnet",     quiet=True)
nltk.download("averaged_perceptron_tagger_eng", quiet=True)

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords, wordnet
from nltk.stem import WordNetLemmatizer
import spacy
from transformers import AutoTokenizer
```

```python
print("Loading models...")
nlp_spacy = spacy.load("en_core_web_sm")
lemmatizer = WordNetLemmatizer()
sw_nltk    = set(stopwords.words("english"))
bert_tok   = AutoTokenizer.from_pretrained("bert-base-uncased")
print("✅ All models loaded\n")
```

```python
# ─────────────────────────────────────────────────────────────────
# NLTK PIPELINE
# ─────────────────────────────────────────────────────────────────
def get_wordnet_pos(tag: str) -> str:
    """Map Penn Treebank POS → WordNet POS."""
    mapping = {"J": wordnet.ADJ, "V": wordnet.VERB, "N": wordnet.NOUN, "R": wordnet.ADV}
    return mapping.get(tag[0], wordnet.NOUN)


def preprocess_nltk(text: str, verbose: bool = False) -> list[str]:
    """
    Full NLTK preprocessing pipeline:
    1. Lowercase
    2. Remove punctuation/numbers
    3. Tokenize (word_tokenize)
    4. Remove stopwords
    5. Lemmatize with POS tags
    """
    # Step 1: Lowercase
    text = text.lower()
    if verbose: print(f"  Step 1 - Lowercase:    '{text}'")

    # Step 2: Remove non-alphabetic characters
    text = re.sub(r"[^a-z\s]", " ", text)
    if verbose: print(f"  Step 2 - Clean:        '{text}'")

    # Step 3: Tokenize
    tokens = word_tokenize(text)
    if verbose: print(f"  Step 3 - Tokens:       {tokens}")

    # Step 4: Remove stopwords + short tokens
    tokens = [t for t in tokens if t not in sw_nltk and len(t) > 2]
    if verbose: print(f"  Step 4 - No stopwords: {tokens}")

    # Step 5: POS-aware lemmatization
    pos_tags = nltk.pos_tag(tokens)
    result   = [lemmatizer.lemmatize(tok, get_wordnet_pos(pos)) for tok, pos in pos_tags]
    if verbose: print(f"  Step 5 - Lemmatized:   {result}")

    return result
```

```python
# ─────────────────────────────────────────────────────────────────
# SPACY PIPELINE
# ─────────────────────────────────────────────────────────────────
def preprocess_spacy(text: str, verbose: bool = False) -> list[str]:
    """
    Full spaCy preprocessing pipeline:
    spaCy does tokenize + POS + lemmatize in a single nlp() call.
    """
    if verbose: print(f"  Input:   '{text}'")

    doc = nlp_spacy(text)

    if verbose:
        print(f"  {'Token':<15} {'is_stop':<10} {'is_alpha':<10} {'lemma_'}")
        for t in doc:
            if t.is_alpha:
                print(f"  {t.text:<15} {str(t.is_stop):<10} {str(t.is_alpha):<10} {t.lemma_}")

    result = [
        token.lemma_.lower()
        for token in doc
        if token.is_alpha         # Only real words
        and not token.is_stop     # Remove stopwords
        and len(token.text) > 2   # Remove very short tokens
    ]

    if verbose: print(f"  Result:  {result}")
    return result
```

```python
# ─────────────────────────────────────────────────────────────────
# HUGGINGFACE PIPELINE (for BERT-style models)
# ─────────────────────────────────────────────────────────────────
def preprocess_hf(text: str, verbose: bool = False) -> dict:
    """
    HuggingFace tokenization pipeline.
    Note: BERT-style models usually keep stopwords —
    they rely on full context including function words.
    """
    if verbose:
        tokens = bert_tok.tokenize(text)
        print(f"  Input:       '{text}'")
        print(f"  Tokens:      {tokens}")
        print(f"  Num tokens:  {len(tokens)} subword tokens")

    encoded = bert_tok(
        text,
        return_tensors="pt",
        padding=True,
        truncation=True,
        max_length=128,
    )

    if verbose:
        ids    = encoded["input_ids"][0].tolist()
        tokens = bert_tok.convert_ids_to_tokens(ids)
        print(f"  Input IDs:   {ids}")
        print(f"  Tokens+spec: {tokens}")

    return {
        "input_ids":      encoded["input_ids"],
        "attention_mask": encoded["attention_mask"],
        "token_type_ids": encoded.get("token_type_ids"),
    }
```

```python
# ─────────────────────────────────────────────────────────────────
# DEMO RUNNER
# ─────────────────────────────────────────────────────────────────

# Test texts
texts = [
  "Machine learning algorithms are being used to analyse large datasets quickly and efficiently.",
  "The researchers studied various methods for improving natural language understanding in AI systems.",
  "Running multiple experiments helped the scientists identify better approaches to data preprocessing.",
]

# ── Show step-by-step on one text ────────────────────────────
print("=" * 65)
print("DEMO: Step-by-Step Preprocessing (verbose mode)")
print("=" * 65)

demo_text = "The running scientists quickly studied better machine learning methods!"
print(f"\nInput: '{demo_text}'\n")

print("── NLTK Pipeline ──")
nltk_result = preprocess_nltk(demo_text, verbose=True)
print(f"  FINAL: {nltk_result}\n")

print("── spaCy Pipeline ──")
spacy_result = preprocess_spacy(demo_text, verbose=True)
print(f"  FINAL: {spacy_result}\n")

print("── HuggingFace (BERT) Pipeline ──")
hf_result = preprocess_hf(demo_text, verbose=True)
print()

# ── Compare all three on multiple texts ──────────────────────
print("=" * 65)
print("COMPARISON: All Three Pipelines on Multiple Texts")
print("=" * 65)

for i, text in enumerate(texts, 1):
  print(f"\n[{i}] Input: '{text}'")
  nltk_r  = preprocess_nltk(text)
  spacy_r = preprocess_spacy(text)
  hf_enc  = bert_tok(text, return_tensors="pt", truncation=True, max_length=64)
  hf_n    = hf_enc["input_ids"].shape[1]
  print(f"  NLTK  ({len(nltk_r):2} tokens) : {nltk_r}")
  print(f"  spaCy ({len(spacy_r):2} tokens) : {spacy_r}")
  print(f"  HF    ({hf_n:2} tokens) : {bert_tok.convert_ids_to_tokens(hf_enc['input_ids'][0][:12])}...")

# ── Token count statistics ────────────────────────────────────
print("\n" + "=" * 65)
print("STATISTICS: Token Count Before vs After Preprocessing")
print("=" * 65)

for text in texts:
  raw_count   = len(text.split())
  nltk_count  = len(preprocess_nltk(text))
  spacy_count = len(preprocess_spacy(text))
  hf_count    = bert_tok(text, return_tensors="pt")["input_ids"].shape[1]

  nltk_pct  = (1 - nltk_count  / raw_count) * 100
  spacy_pct = (1 - spacy_count / raw_count) * 100

  print(f"\n  Text: '{text[:55]}...'")
  print(f"  Raw words:   {raw_count:3}")
  print(f"  NLTK tokens: {nltk_count:3}  ({nltk_pct:.0f}% reduction)")
  print(f"  spaCy tokens:{spacy_count:3}  ({spacy_pct:.0f}% reduction)")
  print(f"  HF tokens:   {hf_count:3}  (subword — can be MORE than raw words)")

print("\n" + "=" * 65)
print("✅ Demo 05 complete — Full Preprocessing Pipeline")
print("=" * 65)
```

DEMO 07 — Real Dataset Preprocessing (IMDb Reviews)

```python
import re
import time
import nltk

nltk.download("punkt",       quiet=True)
nltk.download("punkt_tab",   quiet=True)
nltk.download("stopwords",   quiet=True)
nltk.download("wordnet",     quiet=True)
nltk.download("averaged_perceptron_tagger_eng", quiet=True)

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords, wordnet
from nltk.stem import WordNetLemmatizer
from datasets import load_dataset
from transformers import AutoTokenizer
import spacy
```

```python
print("Loading models and dataset...")
nlp_spacy  = spacy.load("en_core_web_sm", disable=["ner", "parser"])  # Faster: only need tagger+lemmatizer
lemmatizer = WordNetLemmatizer()
sw_nltk    = set(stopwords.words("english"))
bert_tok   = AutoTokenizer.from_pretrained("bert-base-uncased")

# Load a small slice of IMDb dataset (first 200 reviews)
print("Downloading IMDb dataset (first 200 reviews)...")
dataset = load_dataset("stanfordnlp/imdb", "plain_text", split="train[:200]")
print(f"✅ Loaded {len(dataset)} IMDb reviews\n")
```

```python
# ── Helper functions ──────────────────────────────────────────────
def clean_text(text: str) -> str:
    """Basic text cleaning: remove HTML, extra whitespace."""
    text = re.sub(r"<[^>]+>", " ", text)           # Remove HTML tags
    text = re.sub(r"http\S+|www\.\S+", " ", text)  # Remove URLs
    text = re.sub(r"\s+", " ", text).strip()        # Collapse whitespace
    return text.lower()


def get_wordnet_pos(tag: str) -> str:
    m = {"J": wordnet.ADJ, "V": wordnet.VERB, "N": wordnet.NOUN, "R": wordnet.ADV}
    return m.get(tag[0], wordnet.NOUN)


def preprocess_nltk(text: str) -> list[str]:
    text   = re.sub(r"[^a-z\s]", " ", clean_text(text))
    tokens = word_tokenize(text)
    tokens = [t for t in tokens if t not in sw_nltk and len(t) > 2]
    tags   = nltk.pos_tag(tokens)
    return [lemmatizer.lemmatize(t, get_wordnet_pos(p)) for t, p in tags]


def preprocess_spacy_batch(texts: list[str], batch_size: int = 32) -> list[list[str]]:
    """Process multiple texts efficiently using spaCy's pipe()."""
    cleaned = [clean_text(t) for t in texts]
    results = []
    for doc in nlp_spacy.pipe(cleaned, batch_size=batch_size):
        tokens = [
            tok.lemma_.lower()
            for tok in doc
            if tok.is_alpha and not tok.is_stop and len(tok.text) > 2
        ]
        results.append(tokens)
    return results
```

```python
# ── SECTION 1: Sample Reviews ──────────────────────────────────────
print("=" * 65)
print("SECTION 1: Sample IMDb Reviews")
print("=" * 65)

label_map = {0: "NEGATIVE ❌", 1: "POSITIVE ✅"}

for i in range(3):
    review = dataset[i]
    print(f"\n[{i+1}] Label: {label_map[review['label']]}")
    print(f"  Raw text (first 200 chars): '{review['text'][:200]}...'\n")
```

```python
# ── SECTION 2: NLTK Pipeline on 10 reviews ───────────────────────
print("\n" + "=" * 65)
print("SECTION 2: NLTK Pipeline on 10 Reviews")
print("=" * 65)

sample_10 = dataset[:10]
print(f"\n{'#':<4} {'Label':<12} {'Raw words':<12} {'After preproc':<14} {'Sample tokens'}")
print(f"  {'-'*75}")

nltk_start = time.time()
for i, (text, label) in enumerate(zip(sample_10["text"], sample_10["label"])):
    tokens     = preprocess_nltk(text)
    raw_words  = len(text.split())
    reduction  = (1 - len(tokens) / raw_words) * 100
    print(f"  {i+1:<4} {label_map[label]:<12} {raw_words:<12} {len(tokens):<6} (-{reduction:.0f}%)   {tokens[:6]}...")

nltk_time = time.time() - nltk_start
print(f"\n  ⏱️  NLTK time for 10 reviews: {nltk_time:.2f}s")
```

```python
# ── SECTION 3: spaCy Batch Pipeline ──────────────────────────────
print("\n" + "=" * 65)
print("SECTION 3: spaCy Batch Pipeline (50 reviews, using pipe())")
print("=" * 65)

sample_50 = dataset[:50]

spacy_start = time.time()
spacy_results = preprocess_spacy_batch(sample_50["text"])
spacy_time    = time.time() - spacy_start

print(f"\n  Processed 50 reviews in {spacy_time:.2f}s using nlp.pipe()")
print(f"  Average tokens per review: {sum(len(r) for r in spacy_results) / len(spacy_results):.1f}")
print(f"  Average reduction: {100*(1 - sum(len(r) for r in spacy_results) / sum(len(t.split()) for t in sample_50['text'])):.0f}%")

print(f"\n  Sample output (first 5 reviews):")
for i, (tokens, label) in enumerate(zip(spacy_results[:5], sample_50["label"][:5])):
    print(f"  [{i+1}] {label_map[label]}: {tokens[:8]}...")
```

```python
# ── SECTION 4: HuggingFace Tokenizer (BERT) ──────────────────────
print("\n" + "=" * 65)
print("SECTION 4: HuggingFace BERT Tokenizer on 10 Reviews")
print("=" * 65)

sample_texts = dataset["text"][:10]
sample_labels= dataset["label"][:10]

hf_start = time.time()
hf_encoded = bert_tok(
    [clean_text(t) for t in sample_texts],
    padding=True,
    truncation=True,
    max_length=128,
    return_tensors="pt",
)
hf_time = time.time() - hf_start

print(f"\n  Batch tokenized 10 reviews in {hf_time:.3f}s")
print(f"  Input shape: {list(hf_encoded['input_ids'].shape)}")
print(f"    → 10 reviews, each padded/truncated to 128 tokens\n")

print(f"  {'#':<4} {'Label':<12} {'Real tokens':<14} {'Padding tokens'}")
print(f"  {'-'*55}")
for i, (ids, mask, label) in enumerate(zip(
    hf_encoded["input_ids"],
    hf_encoded["attention_mask"],
    sample_labels
)):
    real_toks = mask.sum().item()
    pad_toks  = 128 - real_toks
    print(f"  {i+1:<4} {label_map[label]:<12} {real_toks:<14} {pad_toks} tokens padded")
```

```python
# ── SECTION 5: Token frequency analysis ──────────────────────────
print("\n" + "=" * 65)
print("SECTION 5: Most Common Tokens After Preprocessing")
print("=" * 65)

from collections import Counter

all_results = preprocess_spacy_batch(dataset["text"])
pos_tokens  = []
neg_tokens  = []

for tokens, label in zip(all_results, dataset["label"]):
    if label == 1:
        pos_tokens.extend(tokens)
    else:
        neg_tokens.extend(tokens)

pos_counter = Counter(pos_tokens)
neg_counter = Counter(neg_tokens)

print(f"\n  Top 15 tokens in POSITIVE reviews:")
print(f"  {pos_counter.most_common(15)}")

print(f"\n  Top 15 tokens in NEGATIVE reviews:")
print(f"  {neg_counter.most_common(15)}")

# Tokens unique to each sentiment
pos_set = set(t for t, c in pos_counter.most_common(50))
neg_set = set(t for t, c in neg_counter.most_common(50))

print(f"\n  Tokens dominant in POSITIVE (not in neg top-50): {pos_set - neg_set}")
print(f"  Tokens dominant in NEGATIVE (not in pos top-50): {neg_set - pos_set}")
```

```python
"""
Direct comparison of both libraries on identical inputs.
Shows token count, quality, speed, and practical differences.
Best run interactively — enter your own text to compare live.
"""

import time
import re
import nltk
nltk.download("punkt",     quiet=True)
nltk.download("punkt_tab", quiet=True)
nltk.download("stopwords", quiet=True)
nltk.download("wordnet",   quiet=True)
nltk.download("averaged_perceptron_tagger_eng", quiet=True)

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords, wordnet
from nltk.stem import WordNetLemmatizer, PorterStemmer
import spacy

nlp        = spacy.load("en_core_web_sm")
lemmatizer = WordNetLemmatizer()
porter     = PorterStemmer()
sw_nltk    = set(stopwords.words("english"))
sw_spacy   = nlp.Defaults.stop_words
print("✅ Libraries loaded\n")


# ── Full pipelines ────────────────────────────────────────────────
def get_wordnet_pos(tag):
    m = {"J": wordnet.ADJ, "V": wordnet.VERB, "N": wordnet.NOUN, "R": wordnet.ADV}
    return m.get(tag[0], wordnet.NOUN)


def pipeline_nltk(text: str) -> dict:
    t0     = time.perf_counter()
    lower  = text.lower()
    clean  = re.sub(r"[^a-z\s]", " ", lower)
    tokens = word_tokenize(clean)
    no_sw  = [t for t in tokens if t not in sw_nltk and len(t) > 2]
    pos    = nltk.pos_tag(no_sw)
    lemmas = [lemmatizer.lemmatize(tok, get_wordnet_pos(p)) for tok, p in pos]
    stems  = [porter.stem(t) for t in no_sw]
    sents  = sent_tokenize(text)
    elapsed = (time.perf_counter() - t0) * 1000
    return {
        "library":    "NLTK",
        "sentences":  sents,
        "tokens":     tokens,
        "no_sw":      no_sw,
        "lemmas":     lemmas,
        "stems":      stems,
        "time_ms":    elapsed,
    }


def pipeline_spacy(text: str) -> dict:
    t0  = time.perf_counter()
    doc = nlp(text)
    sents  = [s.text for s in doc.sents]
    tokens = [t.text for t in doc if t.is_alpha]
    no_sw  = [t.text.lower() for t in doc if t.is_alpha and not t.is_stop and len(t.text) > 2]
    lemmas = [t.lemma_.lower() for t in doc if t.is_alpha and not t.is_stop and len(t.text) > 2]
    pos_tags = [(t.text, t.pos_, t.dep_) for t in doc if t.is_alpha]
    elapsed = (time.perf_counter() - t0) * 1000
    return {
        "library":   "spaCy",
        "sentences": sents,
        "tokens":    tokens,
        "no_sw":     no_sw,
        "lemmas":    lemmas,
        "pos_tags":  pos_tags,
        "time_ms":   elapsed,
    }


def print_comparison(text: str):
    """Print a full side-by-side comparison for a given text."""
    print(f"\n{'=' * 70}")
    print(f"Input: '{text}'")
    print(f"{'=' * 70}")

    r_nltk  = pipeline_nltk(text)
    r_spacy = pipeline_spacy(text)

    rows = [
        ("Library",       r_nltk["library"],              r_spacy["library"]),
        ("Sentences",     str(len(r_nltk["sentences"])),  str(len(r_spacy["sentences"]))),
        ("Raw tokens",    str(len(r_nltk["tokens"])),     str(len(r_spacy["tokens"]))),
        ("After stopwords", str(len(r_nltk["no_sw"])),   str(len(r_spacy["no_sw"]))),
        ("After lemma",   str(len(r_nltk["lemmas"])),     str(len(r_spacy["lemmas"]))),
        ("Time (ms)",     f"{r_nltk['time_ms']:.1f}",     f"{r_spacy['time_ms']:.1f}"),
    ]

    print(f"\n  {'Metric':<20} {'NLTK':<25} {'spaCy'}")
    print(f"  {'-'*65}")
    for label, nltk_val, spacy_val in rows:
        print(f"  {label:<20} {nltk_val:<25} {spacy_val}")

    print(f"\n  NLTK  lemmas : {r_nltk['lemmas']}")
    print(f"  NLTK  stems  : {r_nltk['stems']}")
    print(f"  spaCy lemmas : {r_spacy['lemmas']}")

    # POS tags from spaCy (not available directly in NLTK without WordNet)
    print(f"\n  spaCy POS+dependency tags:")
    print(f"  {'Token':<15} {'POS':<10} {'Dep'}")
    for tok, pos, dep in r_spacy["pos_tags"][:10]:
        print(f"  {tok:<15} {pos:<10} {dep}")


# ── Run comparisons ───────────────────────────────────────────────
if __name__ == "__main__":

    test_cases = [
        "The running scientists studied better machine learning methods.",
        "She doesn't like the beautifully designed algorithms.",
        "Dr. Smith's laboratory produced amazing results in 2024.",
        "These wolves are hungrily chasing geese across the frozen lakes.",
        "I'm going to New York to study artificial intelligence next year.",
    ]

    print("=" * 70)
    print("NLTK vs spaCy — Full Pipeline Comparison")
    print("=" * 70)

    for text in test_cases:
        print_comparison(text)

    # ── Speed benchmark ──────────────────────────────────────────
    print("\n" + "=" * 70)
    print("SPEED BENCHMARK: 100 texts processed")
    print("=" * 70)

    benchmark_texts = test_cases * 20  # 100 texts

    # NLTK benchmark
    t0 = time.perf_counter()
    for t in benchmark_texts:
        pipeline_nltk(t)
    nltk_total = (time.perf_counter() - t0) * 1000

    # spaCy benchmark (using pipe for batch efficiency)
    t0 = time.perf_counter()
    for t in benchmark_texts:
        pipeline_spacy(t)
    spacy_total = (time.perf_counter() - t0) * 1000

    # spaCy batch (the right way)
    t0 = time.perf_counter()
    docs = list(nlp.pipe(benchmark_texts, batch_size=32))
    spacy_batch = (time.perf_counter() - t0) * 1000

    print(f"""
  100 texts processed:
  NLTK  (sequential) : {nltk_total:.0f}ms  ({nltk_total/100:.1f}ms/text)
  spaCy (sequential) : {spacy_total:.0f}ms  ({spacy_total/100:.1f}ms/text)
  spaCy (batched)    : {spacy_batch:.0f}ms   ({spacy_batch/100:.1f}ms/text)  ← use nlp.pipe()!

  spaCy is {nltk_total/spacy_total:.1f}x faster than NLTK on this benchmark.
  spaCy batch is {nltk_total/spacy_batch:.1f}x faster when using nlp.pipe().
""")

    # ── Interactive mode ─────────────────────────────────────────
    print("=" * 70)
    print("INTERACTIVE MODE — Enter your own text (Ctrl+C to exit)")
    print("=" * 70)

    try:
        while True:
            user_input = input("\n> Enter text: ").strip()
            if user_input:
                print_comparison(user_input)
    except KeyboardInterrupt:
        print("\n\n✅ Demo 07 complete — NLTK vs spaCy")
```
